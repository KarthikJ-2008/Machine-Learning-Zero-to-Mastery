```python
import pandas as pd
df=pd.read_csv(r"C:\Users\pc\Desktop\Activity\ML\Datasets\Insurance.csv")
print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['bought_insurance'] = le.fit_transform(df['bought_insurance'])
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['age'],df['bought_insurance'])
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(df[['age']], df['bought_insurance'],
                                                    test_size=0.20, random_state=4)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
print("Traing score: ", model.score(x_train, y_train))
y_pred= model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,y_pred))
```
