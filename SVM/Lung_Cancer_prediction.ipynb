```python
#Loading Dataset
import pandas as pd
df=pd.read_csv(r"C:\Users\SPTINT-20\Downloads\survey_lung_cancer_SVM.csv")
df.info()
```

```python
df.head()
```

```python
# Selecting input and output features
x=df.drop(['LUNG_CANCER'],axis=1)
y=df['LUNG_CANCER']
```

```python
# Plotting corretion graph
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 8))
corr=df.corr(numeric_only=True)
sns.heatmap(corr,annot=True,fmt='.2f', cmap='coolwarm')
```

```python
# splitting Dataset
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# importing Support Vector Machine
from sklearn.svm import SVC
model=SVC()
```

```python
# Training model
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix
print("Confusion Matrics:\n",confusion_matrix(y_test,y_pred))
print("accuracy_score:",accuracy_score(y_test,y_pred)*100)
```
