```python
import pandas as pd
df= pd.read_csv(r"C:\Users\pc\Desktop\Activity\ML\Datasets\Mall_Customers _kmeans.csv")
df
```

```python
x=df.iloc[:,[3,4]].values
```

```python
from sklearn.cluster import KMeans

wcss_list=[]
```

```python
for i in range(2,11):
    kmeans = KMeans(n_clusters=i, init='k-means++',random_state=42)
    kmeans.fit(x)
    wcss_list.append(kmeans.inertia_)
```

```python
import matplotlib.pyplot as plt
plt.plot(range(2,11), wcss_list)
plt.title("Elbow Method Graph")
plt.xlabel("number of clusters")
plt.ylabel("wcss_list")
```

```python
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42)
y_pred = kmeans.fit_predict(x)
```

```python
plt.scatter(x[y_pred==0,0], x[y_pred==0,1],s=50,c='green',label='cluster 1')
plt.scatter(x[y_pred==1,0], x[y_pred==1,1],s=50,c='red',label='cluster 2')
plt.scatter(x[y_pred==2,0], x[y_pred==2,1],s=50,c='yellow',label='cluster 3')
plt.scatter(x[y_pred==3,0], x[y_pred==3,1],s=50,c='purple',label='cluster 4')
plt.scatter(x[y_pred==4,0], x[y_pred==4,1],s=50,c='cyan',label='cluster 5')

plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1],s=100, c='black',label='Centroid')

plt.title('Clusters of customers')
plt.xlabel('Annual Income')
plt.ylabel("Spending score")
plt.legend()
```
