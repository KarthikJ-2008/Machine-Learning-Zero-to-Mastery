```python
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\SPTINT-12\Downloads\Insurance.csv")
df
```

```python
from sklearn.linear_model import LinearRegression
```

```python
x=df[['Age']]
y=df['Premium']
```

```python
model=LinearRegression()
```

```python
model.fit(x,y)
```

```python
y_pred=model.predict(x)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y, y_pred)
```

```python
print("MAE  :", mae)
print("MSE  :", mse)
print("RMSE :", rmse)
print("R²   :", r2)
```
